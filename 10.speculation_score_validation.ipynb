```python
import re
import math
from collections import Counter
import pandas as pd

# ---------------------------
# 1. Extract texts from stringified tweets_struct
# ---------------------------
def extract_texts(s):
    if isinstance(s, str):
        # regex to capture everything inside 'text': '...'
        return re.findall(r"'text':\s*'([^']*)'", s)
    elif isinstance(s, list):
        # already a list of dicts
        return [tw["text"] for tw in s]
    else:
        return []

# ---------------------------
# 2. Shannon entropy helpers
# ---------------------------
def shannon_entropy(tokens):
    n = len(tokens)
    if n == 0:
        return 0
    counts = Counter(tokens)
    probs = [c/n for c in counts.values()]
    return -sum(p * math.log2(p) for p in probs)

def char_entropy(texts):
    chars = list("".join(t.replace(" ", "") for t in texts))
    return shannon_entropy(chars)

def word_entropy(texts):
    words = []
    for t in texts:
        words.extend(t.split())
    return shannon_entropy(words)

# ---------------------------
# 3. Feature computation for each row
# ---------------------------
def compute_features(tweet_struct_str):
    texts = extract_texts(tweet_struct_str)
    num_chars = sum(len(t.replace(" ", "")) for t in texts)
    num_words = sum(len(t.split()) for t in texts)
    ce = char_entropy(texts)
    we = word_entropy(texts)
    return pd.Series({
        "num_chars": num_chars,
        "num_words": num_words,
        "char_entropy": ce,
        "word_entropy": we
    })

df_features = df["tweets_struct"].apply(compute_features)
df = pd.concat([df, df_features], axis=1)

df.head()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

def run_regression(df, feature, target):
    # Drop rows with NaN in target/feature
    data = df[[feature, target]].dropna()
    X = data[[feature]]
    y = data[target]
    
    model = LinearRegression().fit(X, y)
    coef = model.coef_[0]
    intercept = model.intercept_
    r2 = model.score(X, y)
    
    print(f"Regression: {target} ~ {feature}")
    print(f"  Coefficient: {coef:.4f}")
    print(f"  Intercept: {intercept:.4f}")
    print(f"  R²: {r2:.4f}")
    print("-" * 40)

# List of features and speculation targets
features = ["num_words", "num_chars", "char_entropy", "word_entropy"]
targets = ["Belief_inpool_Speculation", "Belief_outpool_Speculation"]

# Run regressions
for f in features:
    for t in targets:
        run_regression(df, f, t)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

def run_regression_and_plot(df, feature, target):
    # Drop rows with missing values
    data = df[[feature, target]].dropna()
    X = data[[feature]]
    y = data[target]
    
    # Fit regression
    model = LinearRegression().fit(X, y)
    coef = model.coef_[0]
    intercept = model.intercept_
    r2 = model.score(X, y)
    
    # Print stats
    print(f"Regression: {target} ~ {feature}")
    print(f"  Coefficient: {coef:.4f}")
    print(f"  Intercept: {intercept:.4f}")
    print(f"  R²: {r2:.4f}")
    print("-" * 40)
    
    # Plot scatter + regression line
    plt.figure(figsize=(6, 4))
    plt.scatter(X, y, alpha=0.6, label="Data")
    
    # Line
    x_range = np.linspace(X.min()[0], X.max()[0], 100).reshape(-1, 1)
    y_pred = model.predict(x_range)
    plt.plot(x_range, y_pred, color="red", linewidth=2, label="Regression line")
    
    plt.xlabel(feature)
    plt.ylabel(target)
    plt.title(f"{target} vs {feature}")
    plt.legend()
    plt.show()

# Features and targets
features = ["num_words", "num_chars", "char_entropy", "word_entropy"]
targets = ["Belief_inpool_Speculation", "Belief_outpool_Speculation"]

# Run for all pairs
for f in features:
    for t in targets:
        run_regression_and_plot(df, f, t)
```
