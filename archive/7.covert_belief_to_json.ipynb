```python
import pandas as pd
import json

def as_paragraphs(stance, explanation, speculation, author):
    return [
        {"author": author, "text": f"Stance: {stance}"},
        {"author": author, "text": f"Explanation: {explanation}"},
        {"author": author, "text": f"Speculation: {speculation}"}
    ]

df = pd.read_csv("prompts_medium_beliefs_all_models_by_row.csv")
tasks = []

for i in range(0, len(df), 3):
    row_a = df.iloc[i]
    row_b = df.iloc[i+1]
    row_c = df.iloc[i+2]

    belief_divorce = "OPINION ON WHETHER DIVORCE IN THE USA SHOULD BE EASIER OR MORE DIFFICULT TO OBTAIN THAN IT IS"
    belief_israel = "OPINION ON WHICH SIDE HAS MORE VALID REASONS IN THE ISRAEL‑HAMAS WAR"

    # Divorce task
    tasks.append({
        "belief": belief_divorce,
        "model_a_output": as_paragraphs(row_a["Belief_Divorce_Category"],
                                        row_a["Belief_Divorce_Explanation"],
                                        row_a["Belief_Divorce_Speculation"],
                                        "Model A"),
        "model_b_output": as_paragraphs(row_b["Belief_Divorce_Category"],
                                        row_b["Belief_Divorce_Explanation"],
                                        row_b["Belief_Divorce_Speculation"],
                                        "Model B"),
        "model_c_output": as_paragraphs(row_c["Belief_Divorce_Category"],
                                        row_c["Belief_Divorce_Explanation"],
                                        row_c["Belief_Divorce_Speculation"],
                                        "Model C"),
        "domain": "Divorce"
    })

    # Israel-Hamas task
    tasks.append({
        "belief": belief_israel,
        "model_a_output": as_paragraphs(row_a["Belief_IsraelHamas_Category"],
                                        row_a["Belief_IsraelHamas_Explanation"],
                                        row_a["Belief_IsraelHamas_Speculation"],
                                        "Model A"),
        "model_b_output": as_paragraphs(row_b["Belief_IsraelHamas_Category"],
                                        row_b["Belief_IsraelHamas_Explanation"],
                                        row_b["Belief_IsraelHamas_Speculation"],
                                        "Model B"),
        "model_c_output": as_paragraphs(row_c["Belief_IsraelHamas_Category"],
                                        row_c["Belief_IsraelHamas_Explanation"],
                                        row_c["Belief_IsraelHamas_Speculation"],
                                        "Model C"),
        "domain": "IsraelHamas"
    })

with open("prompts_medium_beliefs_all_models_by_row.json", "w", encoding="utf-8") as f:
    json.dump(tasks, f, indent=2, ensure_ascii=False)

print("Tasks written to prompts_medium_beliefs_all_models_by_row.json")
```
