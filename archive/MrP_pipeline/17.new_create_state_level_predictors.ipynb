```python
def add_variables(variable_code, new_name, df, strat_frame):
    df = df[df[variable_code] >= 0].copy()
    df_agg = df.groupby('state_simple')[variable_code].mean().reset_index()
    df_agg[new_name] = stats.zscore(df_agg[variable_code])
    strat_frame = strat_frame.merge(df_agg[['state_simple', new_name]], on='state_simple', how='left')
    return strat_frame
```

in mapper, for all those with Reverse column = True(1), take the Match ID numeric and find in the anes dataset column[Match ID numeric], revert all the positive values in those columns (e.g, if the postive values ranges from 1-5, then replace 1 with 5, 2 with 4, 3 stays the same, 4 with 2 and 5 with 1). Don't put any error wrapper and make sure to raise the error and stop the code whenever any issue show up. After the operation print me a few examples to ensure that it's done correctly.

```python
def reverse_code_from_mapper(match_dataset):
    # pick source dataframe based on Match Dataset
    dataset_key = str(match_dataset).strip().upper()
    dataset_to_df = {
        "ANES2020": anes,
        "ANES2022": anes2022
    }
    df = dataset_to_df[dataset_key]  # raises KeyError if dataset not found

    reverse_mask = (
        mapper["Reverse"]
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(["1", "1.0", "true"])
        &
        mapper["Match Dataset"]
        .astype(str)
        .str.strip()
        .str.upper()
        .eq(dataset_key)
    )

    reverse_rows = mapper.loc[reverse_mask, ["Match ID numeric", "Reverse", "Match Dataset"]].copy()
    reverse_cols = reverse_rows["Match ID numeric"].astype(str).str.strip().unique()

    n_reverse_cols = len(reverse_cols)
    print(f"[{dataset_key}] Number of columns to reverse: {n_reverse_cols}")

    original_values = {}

    for col in reverse_cols:
        s = df[col]
        pos_mask = s > 0

        if not pos_mask.any():
            raise ValueError(f"{dataset_key}: column '{col}' has no positive values to reverse-code.")

        min_pos = s.loc[pos_mask].min()
        max_pos = s.loc[pos_mask].max()

        original_values[col] = s.copy()
        df.loc[pos_mask, col] = (min_pos + max_pos) - s.loc[pos_mask]

    # mark processed mapper rows
    mapper.loc[reverse_mask, "Reverse"] = 0

    print(f"[{dataset_key}] Reverse-coded {n_reverse_cols} columns.")
    print(f"[{dataset_key}] Set Reverse=0 for {reverse_mask.sum()} mapper rows.")

    # preview
    for col in reverse_cols[:5]:
        before = original_values[col]
        after = df[col]
        pos_before = before[before > 0]

        min_pos = pos_before.min()
        max_pos = pos_before.max()

        print(f"\n[{dataset_key}] Column: {col} | positive range reversed: [{min_pos}, {max_pos}]")

        mapping_preview_vals = np.sort(pos_before.dropna().unique())
        mapping_preview = pd.DataFrame({
            "before": mapping_preview_vals,
            "after": (min_pos + max_pos) - mapping_preview_vals
        })
        print("Value mapping preview:")
        print(mapping_preview.head(10))

        changed = (before > 0) & (before != after)
        print("Row-level before/after preview:")
        print(pd.DataFrame({"before": before[changed], "after": after[changed]}).head(5))

    return reverse_cols
```
