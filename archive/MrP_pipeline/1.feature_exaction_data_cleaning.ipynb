```python
# unify the symbols output format
def clean_column_pattern(value, row_idx, col_name):
    if pd.isna(value):
        return value
    value_str = str(value)

    match = re.search(r'([a-zA-Z]+\d+)\)?', value_str)
    if match:
        return match.group(1)
    print(f"Row {row_idx}, Column '{col_name}': Unmatched value '{value_str}'")
    return value

columns_to_clean = [col for col in df.columns if '_Symbol' in col and col != 'Demo_which_state_of_the_usa_do_they_live_in__Symbol']
for col in columns_to_clean:
    df[col] = pd.Series(
        (clean_column_pattern(val, row_idx=idx, col_name = col) for idx, val in df[col].items()),
        index = df.index
    )
```

# standarlizing

```python
# Standardize letter prefixes in symbol columns (sometimes there are a multiple letter prefix in one column)
for col in columns_to_clean:
    non_null_values = df[col].dropna()

    if len(non_null_values) == 0:
        print(f"Column '{col}' has no non-null values.\n")
        continue

    letter_prefixes = set()
    prefix_counts = {}
    for idx, val in non_null_values.items():
        match = re.match(r'([a-zA-Z]+)', str(val))
        if match:
            prefix = match.group(1)
            letter_prefixes.add(prefix)
            prefix_counts[prefix] = prefix_counts.get(prefix, 0) + 1

    if len(letter_prefixes) > 1:
        print(f"Column '{col}' has multiple letter prefixes: {letter_prefixes}")
        for prefix, count in sorted(prefix_counts.items()):
            print(f"  {prefix}: {count}")
        # Find the most common prefix
        most_common_prefix = max(prefix_counts, key=prefix_counts.get)

        df[col] = df[col].apply(
            lambda x: re.sub(r'^[a-zA-Z]+', most_common_prefix, str(x)) if pd.notna(x) else x
        )
        print()
```

# here i want to build a seperate dataframe that has one column that contains all the belief names (for example 'Belief_opinion_on_declining_share_of_white_people_in_america_Symbol'), and another column contains the most common letter prefix (e.g. rab)

```python
# Build a summary of belief symbol columns and their most common letter prefix
belief_symbol_cols = [c for c in df.columns if c.startswith("Belief_") and c.endswith("_Symbol")]

def most_common_prefix(series):
    prefixes = (
        series.dropna()
        .astype(str)
        .str.extract(r"^([a-zA-Z]+)", expand=False)
        .dropna()
    )
    if prefixes.empty:
        return pd.NA
    return prefixes.value_counts().idxmax()

belief_prefix_df = pd.DataFrame({
    "belief_name": belief_symbol_cols,
    "most_common_prefix": [most_common_prefix(df[c]) for c in belief_symbol_cols],
})

```

```python
# Unified state cleanup mapping
STATE_MAPPING = {
    # Standard abbreviations
    'alabama': 'AL', 'alaska': 'AK', 'arizona': 'AZ', 'arkansas': 'AR', 'california': 'CA',
    'colorado': 'CO', 'connecticut': 'CT', 'delaware': 'DE', 'florida': 'FL', 'georgia': 'GA',
    'hawaii': 'HI', 'idaho': 'ID', 'illinois': 'IL', 'indiana': 'IN', 'iowa': 'IA',
    'kansas': 'KS', 'kentucky': 'KY', 'louisiana': 'LA', 'maine': 'ME', 'maryland': 'MD',
    'massachusetts': 'MA', 'michigan': 'MI', 'minnesota': 'MN', 'mississippi': 'MS', 'missouri': 'MO',
    'montana': 'MT', 'nebraska': 'NE', 'nevada': 'NV', 'new hampshire': 'NH', 'new jersey': 'NJ',
    'new mexico': 'NM', 'new york': 'NY', 'north carolina': 'NC', 'north dakota': 'ND', 'ohio': 'OH',
    'oklahoma': 'OK', 'oregon': 'OR', 'pennsylvania': 'PA', 'rhode island': 'RI', 'south carolina': 'SC',
    'south dakota': 'SD', 'tennessee': 'TN', 'texas': 'TX', 'utah': 'UT', 'vermont': 'VT',
    'virginia': 'VA', 'washington': 'WA', 'west virginia': 'WV', 'wisconsin': 'WI', 'wyoming': 'WY',

    # D.C.
    'district of columbia': 'DC', 'washington dc': 'DC', 'dc': 'DC',

    # Comma variations
    'ny,': 'NY', 'wa,': 'WA',
    'california,': 'CA', 'new york,': 'NY', 'texas,': 'TX', 'florida,': 'FL',
    'illinois,': 'IL', 'massachusetts,': 'MA', 'michigan,': 'MI',
    'minnesota,': 'MN', 'arizona,': 'AZ', 'nevada,': 'NV', 'tennessee,': 'TN',
    'south carolina,': 'SC', 'kansas,': 'KS', 'louisiana,': 'LA', 'maine,': 'ME',
    'connecticut,': 'CT', 'utah,': 'UT', 'idaho,': 'ID',

    # Parenthetical variations
    'california (ca)': 'CA', 'arizona (az)': 'AZ', 'kansas (ks)': 'KS',
    'tennessee (tn)': 'TN', 'hi (hawaii)': 'HI', 'ma (massachusetts)': 'MA',
    'ca (california)': 'CA',

    # USA 
    'usa': 'USA', 'usa,': 'USA', 'us':'USA',
    'usa (no specific state)': 'USA',
    'usa (unspecified state)': 'USA',
    'usa (not a specific state)': 'USA',
    'usa (no specific state identified)': 'USA',
    'usa (the user is from the united states but does not specify a state)': 'USA',
    'usa (the user is from the united states but no specific state is identified)': 'USA',
    'if they do not specify a state, but are still from the united states of america, write "usa"': 'USA',

    # No state specified
    'text': 'No_state', 'na': 'No_state', 'n/a': 'No_state', 'xx': 'No_state',
    'not from a state in the usa': 'No_state',
    'not from a state in the usa,': 'No_state',
    'no state': 'No_state',
}

def clean_state(state):
    if pd.isna(state):
        return np.nan
    s = str(state).strip().lower().rstrip(')')
    # Remove 'text)' prefix if present
    if s.startswith('text)'):
        s = s[6:].strip()

    if s in STATE_MAPPING:
        return STATE_MAPPING[s]
    
    if len(s) == 2 and s.upper() in STATE_MAPPING.values():
        return s.upper()    
    
    # If direct mapping fails, try partial matches, longest keys first to avoid false positives
    for key in sorted(STATE_MAPPING.keys(), key=len, reverse=True):
        if key in s:
            return STATE_MAPPING[key]

    print(f"No match found: '{s}'")
    return s.upper()

df['state_cleaned'] = df['Demo_which_state_of_the_usa_do_they_live_in_Symbol'].apply(clean_state)
print(len(df['state_cleaned'].unique()))
print(df['state_cleaned'].value_counts())
```

```python
# add lickert order detection for belief columns
belief_cols = df.columns[df.columns.str.startswith('Belief_') & df.columns.str.endswith('_Symbol')]

def extract_number(cat):
    match = re.search(r"(\d+)$", str(cat))
    if match:
        return int(match.group(1))
    else:
        return None
    
def detect_likert_order(df, col):
    cats = df[col].dropna().unique()
    ordered_cats = sorted(cats, key=extract_number)
    return ordered_cats

for col in belief_cols:
    likert_order = detect_likert_order(df, col)

    df[col + '_cat'] = pd.Categorical(df[col], categories=likert_order, ordered=True)
    df[col + '_num'] = df[col + '_cat'].cat.codes.replace(-1, pd.NA) + 1
```

create a new column in df named Belief_increase_fossil_fuel_production_Symbol, construct it in a way with the following steps: 
1. take the following three columns:
'Belief_opinion_on_increasing_the_production_availability_and_usage_of_energy_from_hydraulic_fracturing_also_known_as_fracking_Symbol_num'
'Belief_opinion_on_increasing_the_production_availability_and_usage_of_energy_from_offshore_oil_and_gas_drilling_Symbol_num'
'Belief_opinion_on_increasing_the_production_availability_and_usage_of_energy_from_coal_mining_Symbol'
2. for each observation, if any of the values from these three columns contains 1 or 2, assign value 1 to column "Belief_increase_fossil_fuel_production_Symbol", if non of the value from the three columns are 1 or 2, assign value 2 to column "Belief_increase_fossil_fuel_production_Symbol", assign also the same value for another column "Belief_increase_fossil_fuel_production_Symbol_num", and "Belief_increase_fossil_fuel_production_Symbol_cat" has the same value but coded as likert scale categories
so 1 - agree increase fossil fuel production, 2- disagree increase fossil fuel production

```python
# Build fossil-fuel-production belief variable with updated logic (NA/1/2)
fracking_col = "Belief_opinion_on_increasing_the_production_availability_and_usage_of_energy_from_hydraulic_fracturing_also_known_as_fracking_Symbol_num"
offshore_col = "Belief_opinion_on_increasing_the_production_availability_and_usage_of_energy_from_offshore_oil_and_gas_drilling_Symbol_num"
coal_col = "Belief_opinion_on_increasing_the_production_availability_and_usage_of_energy_from_coal_mining_Symbol_num"

src = df[[fracking_col, offshore_col, coal_col]].apply(pd.to_numeric, errors="coerce")

all_na = src.isna().all(axis=1)
any_one_two = src.isin([1, 2]).any(axis=1)

score = pd.Series(pd.NA, index=df.index, dtype="Int64")
score.loc[~all_na & any_one_two] = 1
score.loc[~all_na & ~any_one_two] = 2

df["Belief_increase_fossil_fuel_production_Symbol"] = score
df["Belief_increase_fossil_fuel_production_Symbol_num"] = score
df["Belief_increase_fossil_fuel_production_Symbol_cat"] = pd.Categorical(
    score, categories=[1, 2], ordered=True
)
```

```python
import pickle

def encode_var(df, col):
    le = LabelEncoder()
    notna_mask = df[col].notna()
    encoded = pd.Series(np.nan, index=df.index)
    encoded[notna_mask] = le.fit_transform(df.loc[notna_mask, col]) + 1

    mapping = {
        original: int(code + 1)
        for code, original in enumerate(le.classes_)
    }

    return encoded, mapping

predictor_cols = [
    "Demo_sex_Symbol",
    "Demo_age_Symbol",
    "Demo_ethnicity_Symbol",
    "Demo_income_Symbol",
    "Demo_education_Symbol",
    "Demo_past_vote_Symbol"
]

all_mappings = {}

for col in predictor_cols:
    encoded, mapping = encode_var(df, col)
    new_col_name = col.replace('_Symbol', '') + '_idx'
    df[new_col_name] = encoded
    all_mappings[col] = mapping

print(all_mappings)


shp = gpd.read_file("/mnt/e/Dropbox/is_culture_king/is_culture_king_Yuyue/generated_data/tl_2025_us_state/tl_2025_us_state.shp")
shp["STATE"] = shp["STUSPS"].str.upper()

shp = shp[~shp["STATE"].isin(["PR", "GU", "VI", "MP", "AS"])]

shp = shp.sort_values("STATE").reset_index(drop=True)

state_order = shp["STATE"].tolist()

state_to_idx = {s: i+1 for i, s in enumerate(state_order)}

df["state_cleaned_idx"] = (
    df["state_cleaned"]
      .str.upper()
      .map(state_to_idx)
)


with open('/mnt/e/Dropbox/is_culture_king/is_culture_king_Yuyue/generated_data/cleaned_MrP_data/all_mappings.pkl', 'wb') as f:
    pickle.dump(all_mappings, f)
```
