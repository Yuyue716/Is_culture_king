```python
raw_df = pd.read_csv("/mnt/c/Users/yuyue/Dropbox/is_culture_king/is_culture_king_Yuyue/generated_data/feature_extraction/drop_usa_cleaned_all_result.csv")
```

```python
vote_map = {
    "vpa2": 1,   # Trump
    "vpa3": -1,  # Biden
    "vpa1": 0,   # did not vote
    "vpa4": 0,
    "vpa5": 0,
    "vpa6": 0
}

raw_df["vote_ideo"] = raw_df["Demo_past_vote_Symbol"].map(vote_map)
```

```python
belief_map = {
    "deb1":  1,   # insurrection
    "deb2": -1,   # protected protest
    "deb3":  0    # unfortunate but move on
}

raw_df["January_belief_num"] = raw_df[belief_col].map(belief_map)
```

```python
pid_map = {
    "pid1": -3,
    "pid2": -2,
    "pid3": -1,
    "pid4": 0,
    "pid5": 1,
    "pid6": 2,
    "pid7": 3
}

raw_df["pid_score"] = raw_df["Demo_partisan_loyalties_Symbol"].map(pid_map)
```

```python
raw_df["Demo_partisan_loyalties_Symbol"].value_counts()
```

```python
raw_df.head()
```

```python
raw_state = (
    raw_df
    .groupby("state_cleaned")
    .agg(
        raw_belief_mean=(belief_col+"_num", "mean"),
        raw_party_mean=("pid_score", "mean"),
        n_raw=(belief_col+"_num", "size")
    )
    .reset_index()
)
```

```python
state_mapping = raw_df[['state_cleaned_idx', 'state_cleaned']].drop_duplicates().sort_values('state_cleaned_idx')
state_mapping_dict = dict(zip(state_mapping['state_cleaned_idx'], state_mapping['state_cleaned']))

state_preds['state_cleaned'] = state_preds['state_idx'].map(state_mapping_dict)
```

```python
state_df = raw_state.merge(state_preds, left_on="state_cleaned", right_on="state_cleaned", how="inner")
```

```python
state_df.head()
```

```python
from scipy.stats import pearsonr, spearmanr

pearson_r, pearson_p = pearsonr(
    state_df["raw_party_mean"],
    state_df["raw_belief_mean"]
)

spearman_r, spearman_p = spearmanr(
    state_df["raw_party_mean"],
    state_df["raw_belief_mean"]
)

print(f"Pearson r = {pearson_r:.3f} (p={pearson_p:.3g})")
print(f"Spearman ρ = {spearman_r:.3f} (p={spearman_p:.3g})")
```

```python
from scipy.stats import pearsonr, spearmanr

pearson_r, pearson_p = pearsonr(
    state_df["raw_party_mean"],
    state_df["pred_mean"]
)

spearman_r, spearman_p = spearmanr(
    state_df["raw_party_mean"],
    state_df["pred_mean"]
)

print(f"Pearson r = {pearson_r:.3f} (p={pearson_p:.3g})")
print(f"Spearman ρ = {spearman_r:.3f} (p={spearman_p:.3g})")
```

```python
state_df["pid_bin"] = pd.cut(
    state_df["raw_party_mean"],
    bins=[-3.5, -1.5, -0.25, 0.25, 1.5, 3.5],
    labels=[
        "Strong D",
        "Lean D",
        "Independent",
        "Lean R",
        "Strong R"
    ]
)
```

```python
binned = (
    state_df
    .groupby("pid_bin")
    .agg(
        mean_mrp_belief=("pred_mean", "mean"),
        mean_raw_belief=("raw_belief_mean", "mean"),
        n_states=("state_cleaned", "size")
    )
    .reset_index()
)

print(binned)
```
