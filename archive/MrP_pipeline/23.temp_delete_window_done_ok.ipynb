```python
import os
import re

base_dir = r"/mnt/e/Dropbox/is_culture_king/is_culture_king_Yuyue/generated_data"
window_dir_pattern = re.compile(r"^Belief_.*_stan_output$")

for name in os.listdir(base_dir):
    belief_path = os.path.join(base_dir, name)
    if not os.path.isdir(belief_path) or not name.startswith("Belief_"):
        continue

    for window_name in os.listdir(belief_path):
        if not window_dir_pattern.match(window_name):
            continue

        window_path = os.path.join(belief_path, window_name)
        done_path = os.path.join(window_path, "window_done.ok")
        if os.path.exists(done_path):
            os.remove(done_path)
            print(f"Deleted: {done_path}")
```
