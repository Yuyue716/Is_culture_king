# add state vote LP

```python
# aggregate over state_electoral.college for Maine and Nebraska
pct_cols = [c for c in state_vote_LP.columns if c.startswith('pct.2020.votes.')]

def _weighted_pct_2020(group):
    # default weight for most pct columns
    results = {}
    for c in pct_cols:
        if c == 'pct.2020.votes.stay.home' or c == 'pct.2020.votes.T':
            w = group['N.pop.state']
        else:
            w = group['2020.votes.T']
        results[c] = np.average(group[c], weights=w)
    return pd.Series(results)

vote2020 = (
    state_vote_LP
    .groupby('state_simple', dropna=False)
    .apply(_weighted_pct_2020)
    .reset_index()
)

vote2020.head()
```

# add ANES 2020 LPs

```python
def add_anes_variables(variable_code, new_name, anes=anes, strat_frame=strat_frame):
    anes = anes[anes[variable_code] >= 0].copy()
    anes_agg = anes.groupby('state_simple')[variable_code].mean().reset_index()
    anes_agg[new_name] = stats.zscore(anes_agg[variable_code])
    strat_frame = strat_frame.merge(anes_agg[['state_simple', new_name]], on='state_simple', how='left')
    return strat_frame
```

```python
# add trust LP
strat_frame = add_anes_variables('V201237', 'State_Trust_LP',strat_frame=strat_frame)
```

```python
# add immigration LP
strat_frame = add_anes_variables('V202232', 'Belief_opinion_on_the_number_of_immigrants_in_america_Symbol_LP',strat_frame=strat_frame)
```
