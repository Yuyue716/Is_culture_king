```python
sex_mapping = {'M': 1, 'F': 2}  
age_mapping = {'1.18-24': 1, '2.25-34': 2, '3.35-44': 3, '4.45-54': 4, '5.55-64': 5, '6.65+': 6 } 
eth_mapping = {'White': 1, 'Black': 2, 'Hispanic': 3, 'Asian': 4, 'Native': 5, 'Mixed-race': 6 }
income_mapping = {'1.[min, 25000)': 1, '2.[25000, 50000)': 2, '3.[50000, 75000)': 3,  '4.[75000, 100000)': 4, '5.[100000, max]': 5}
vote_mapping = {'stay home': 1, 'R': 2, 'D': 3, 'L': 4, 'G': 5, 'OTHER': 6 }
education_mapping = {'Does not have a College Degree': 1, 'Has a College Degree': 2}
```

```python
strat_df['sex_idx'] = strat_df['gender'].map(sex_mapping)
strat_df['age_idx'] = strat_df['age_bins'].map(age_mapping)
strat_df['eth_idx'] = strat_df['ethnicity'].map(eth_mapping)
strat_df['income_idx'] = strat_df['hh_income'].map(income_mapping)
strat_df['vote_idx'] = strat_df['vote2020'].map(vote_mapping)
strat_df['education_idx'] = strat_df['college_grad'].map(education_mapping)
```

```python
shp = gpd.read_file("/mnt/e/Dropbox/is_culture_king/is_culture_king_Yuyue/generated_data/tl_2025_us_state/tl_2025_us_state.shp")
shp["STATE"] = shp["STUSPS"].str.upper()

shp = shp[~shp["STATE"].isin(["PR", "GU", "VI", "MP", "AS"])]

shp = shp.sort_values("STATE").reset_index(drop=True)

state_order = shp["STATE"].tolist()

state_to_idx = {s: i+1 for i, s in enumerate(state_order)}

strat_df = strat_df[strat_df['state_abbreviation'].isin(state_order)].copy()

# Assign numeric state index according to shapefile order
strat_df['state_idx'] = strat_df['state_abbreviation'].map(state_to_idx)

print(strat_df[['state_abbreviation', 'state_idx']].head())
```

```python
import numpy as np
```

```python
predictor_cols = ['sex_idx', 'age_idx', 'eth_idx', 'income_idx', 'state_idx', 'vote_idx', 'education_idx']

print("Unique values per predictor column:")
for col in predictor_cols:
    print(f"{col}: {strat_df[col].nunique()}")
    
unique_counts = [strat_df[col].nunique() for col in predictor_cols]
total_combinations = np.prod(unique_counts)
print(f"Total possible combinations of predictors: {total_combinations}")

unique_combinations = strat_df[predictor_cols].drop_duplicates().shape[0]
print(f"Number of unique predictor combinations actually present: {unique_combinations}")

print(f"Original number of rows: {len(strat_df)}")
```

```python
predictor_cols = ['sex_idx', 'age_idx', 'eth_idx', 'income_idx', 'state_idx', 'vote_idx', 'education_idx']
n_star_cols = [col for col in strat_df.columns if col.startswith('N_star')]
other_cols = [col for col in strat_df.columns if col not in predictor_cols and col not in n_star_cols]

# Aggregate N_star columns with mean
mean_agg = strat_df.groupby(predictor_cols, as_index=False, observed=True)[n_star_cols].mean()

# Aggregate other columns with first
first_agg = strat_df.groupby(predictor_cols, as_index=False, observed=True)[other_cols].first()

strat_df = mean_agg.merge(first_agg, on=predictor_cols, how='left')
```
