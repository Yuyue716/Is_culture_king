```python
import pandas as pd
import requests


# Function to check if the image URL is valid
def is_image_accessible(url):
    try:
        response = requests.head(url, timeout=5, allow_redirects=True)
        return response.status_code == 200
    except requests.RequestException:
        return False
```

```python
# Apply the function to each profile image URL
print("Checking profile image availability...")
user_profiles["image_accessible"] = user_profiles["profile_image_url"].apply(is_image_accessible)

# Report
n_total = len(user_profiles)
n_missing = (~user_profiles["image_accessible"]).sum()
print(f"{n_missing} out of {n_total} profile images are no longer accessible.")

user_profiles[~user_profiles["image_accessible"]].to_csv("../generated_data/new_unavailable_profile_images.csv", index=False)
```

```python
#merge manually filled images back to the original profile

user_profiles = pd.read_csv("../generated_data/profiles AND metadata")

manual_fixes = pd.read_excel("../generated_data/final_unavailable_profile_images.xlsx")

user_profiles = user_profiles.merge(
    manual_fixes[["gpt_id", "profile_image_url"]],
    on="gpt_id",
    how="left",
    suffixes=("", "_fixed")
)

# If the manually filled profile_image_url_fixed exists, use it, else keep original
user_profiles["profile_image_url"] = user_profiles["profile_image_url_fixed"].combine_first(
    user_profiles["profile_image_url"]
)

# Drop helper column
user_profiles = user_profiles.drop(columns=["profile_image_url_fixed"])

# re-check accessibility of the updated URLs
print("Re-checking manually filled profile image URLs...")
user_profiles["image_accessible"] = user_profiles["profile_image_url"].apply(is_image_accessible)

user_profiles.to_csv("../generated_data/user_profiles_fixed.csv", index=False)
user_profiles
```

```python
import pandas as pd
import requests

def is_url_accessible(url):
    try:
        response = requests.head(url, timeout=5, allow_redirects=True)
        return response.status_code == 200
    except requests.RequestException:
        return False

def remove_twitter_suffix(url):
    if not isinstance(url, str):
        return url, False 
    new_url = url
    if "_normal." in url:
        new_url = url.replace("_normal.", ".")
    elif "_400x400." in url:
        new_url = url.replace("_400x400.", ".")

    if new_url == url:
        return url, False  

    # Check if new URL works, otherwise keep the original
    if is_url_accessible(new_url):
        return new_url, False  
    else:
        return url, True 

# Apply function to DataFrame
kept_count = 0
new_urls = []
for url in df["profile_image_url"]:
    new_url, kept = remove_twitter_suffix(url)
    if kept:
        kept_count += 1
    new_urls.append(new_url)

df["profile_image_url"] = new_urls

print(f"Finished processing. Kept original URL for {kept_count} rows because cleaned version was inaccessible.")
```
